```python
import os, random, shutil
```

```python
# Paths
raw_path = "/mnt/e/synthetic-data-generation/Helmet_Dataset/"
base_dir = "/mnt/e/synthetic-data-generation/YOLO_Data"

# Create folders
for f in ["train/images", "train/labels", "val/images", "val/labels"]:
    os.makedirs(os.path.join(base_dir, f), exist_ok=True)
```

```python
# Split 80/20
images = [f for f in os.listdir(raw_path) if f.endswith('.png')]
random.shuffle(images)
split = 800

def move_files(files, group):
    for f in files:
        shutil.copy(os.path.join(raw_path, f), os.path.join(base_dir, group, "images", f))
        lbl = f.replace(".png", ".txt")
        if os.path.exists(os.path.join(raw_path, lbl)):
            shutil.copy(os.path.join(raw_path, lbl), os.path.join(base_dir, group, "labels", lbl))
        else:
            open(os.path.join(base_dir, group, "labels", lbl), 'w').close()

move_files(images[:split], "train")
move_files(images[split:], "val")
print("Data Split Done.")
```

```python
yaml_text = f"""
path: {base_dir}
train: train/images
val: val/images
names: 
    0: Helmet
"""
with open("data.yaml", "w") as f:
    f.write(yaml_text)
print("Config Created.")
```

```python
from ultralytics import YOLO

# Load Nano model
model = YOLO('yolov8n.pt')

# Start training
model.train(
    data = 'data.yaml',
    epochs = 50,
    imgsz = 640,
    device = 0
)
```
